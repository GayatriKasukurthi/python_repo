```python
# If statement excecutes query when the condition is true
Age = 7
if Age >= 18:
    print("eligible to marry")
```

```python
# If else executes between two conditions
Age = 7
if Age >= 18:
    print("eligible to marry")
else:
    print("not eligible to marry")
```

```python
# if elif else executes between multiple conditions
Age = 13
if Age >= 18:
    print("eligible to marry")
    
elif Age >= 13:
    print("eligible to date")
    
else:
    print("not eligible to marry")
```

```python
# For loop repeat code for each item in a sequence
fruits = ["Apple", "banana","tomato"]
for i in fruits:
    print(i)
```

```python
for (i) in range(6):
    print(i)
```

```python
# while repeat until condition becomes true
count = 5
while count <=2:
    print(count)
    count -=1 # count = Count + 1
```

```python
for i in range(1,11):
    if i == 10:
        break
    print(i)
    
```

```python
# continue skip the current iteration and continue with the next one
for i in range(11):
    if i == 5:
        continue
    print(i)
```

```python
# functions there are two 
#def () in bussiness large logics like health care, banking,insurance.
# Lambda () simple logic, simple filters mappings.
# def means define a function
# test mean name of the function
# () parameters
# : start of the function 
# indentation - code inside the function

def test(): # parameters
    print("welcome to the python learning series")
test() # arguments
# without parameters we can see above
# with parameters we can see below

def test(a, b): # parameters
    print(a+b)
test(3,6) # arguments
```

```python
# positional arguments
def test(Age, name):
    print(Age, "Name:")
    print("name:", Age)
    
test(28, "Ammul")
```

```python
# keyword arguments
def test(Age, name):
    print(Age, "name:")
    print("name:", Age)
    
test(Age = 28, name = "Ammul")
```

```python
# Default parameters
def test(name = "India but i want to go other countires like Canada, USA, Switzerland, Germany, France, Italy, Spain, Portugal, Netherlands, Belgium, Austria, Denmark, Sweden, Norway, Finland, Iceland, Ireland, Scotland, Wales, England"):
    print(name)
       
test("canada")
```

```python
# args(variable number of arguments)
def add(* numbers):
    print(sum(numbers))
add(20,20,50)
```

```python
def clean_salary(salary): # parameters
    if clean_salary is None: # none is special object in python
      return 0
    return salary

test = clean_salary(None) # arguments 
print (test)
```

```python
# keywordargs(variable number of arguments) ** means keyword args
def add(** numbers):
    print(numbers)
    
add(name = "Jhon", age = 28, country = "India")
```

```python
# def function is used when you need to write resuable code that performs a specific task

def add(a,b): # a, b are parameters
    return a+b # logic 10+20 = 40
print(add(10,30)) # arguments # calling
    
```

```python
# lambda function is used to create a small one line function
# short calculation, simple logic, filtering, simple transformations
add = lambda x, y: x+y # lambda parameters: expression # here add is a variable in above def function add is a function
print(add(10,20))
```

```python
add = lambda x, y: x+y # lambda parameters: expression
print(add(10,20))
```

```python
def square(x): # here x is the parameter # here square is the function name
    return x*x # logic 
print(square(5)) # arguments
```

```python
square = lambda x: x*x  # here square is a variable name ## Lambda parameters: experession
print(square(5)) # at the end we need to call variable name here square is the variable name
```
